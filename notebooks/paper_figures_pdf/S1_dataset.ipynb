```python
import torch
import matplotlib.pyplot as plt
import numpy as np
from dinosaw.utils import add_custom_font
import glob
from os import listdir
from PIL import Image


SEED = 100001
torch.manual_seed(SEED)
np.random.seed(SEED)
```

## Microstructures

```python
ds_folder = 'data/linear_probe/homog_micros'
image_files = [f for f in listdir(ds_folder)]
n_imgs = len(image_files)

imgs = []
for img_file in image_files:
    img_path = f'{ds_folder}/{img_file}'
    imgs.append(Image.open(img_path).convert('RGB'))
```

```python
ds_folder = 'data/linear_probe/texture_ds'
image_files = [f for f in listdir(ds_folder)]
np.random.shuffle(image_files)
```

```python
# imgs = []
for img_file in image_files[:15]:
    img_path = f'{ds_folder}/{img_file}'
    imgs.append(Image.open(img_path).convert('RGB'))
```

```python


plt.style.use("thesis.mplstyle")
plt.rcParams['text.usetex'] = False
add_custom_font('resources/fonts', 'Grotesk')

NROW, NCOL = 6, 5
fig, axes = plt.subplots(NROW, NCOL, figsize=(5, 2 * 2.3))
for ax, img in zip(axes.ravel(), imgs):
    ax.imshow(img)
    ax.set_xticks([])
    ax.set_yticks([])


labels = ['(a)', '(b)']
y_off = [1.05, 1.05]
axes = [axes[0, 0], axes[-3, 0]]

for i in range(2):
    ax = axes[i]
    ax.text(-0.4, y_off[i], labels[i], transform=ax.transAxes,
            fontweight='bold', color='black')


SAVE = True
if SAVE:
    plt.savefig('saved/S1.jpeg', dpi=300, bbox_inches='tight')
    plt.close()
```
