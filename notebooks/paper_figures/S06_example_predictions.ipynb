```python
import torch
from dinosaw.helpers import add_custom_font
from dinosaw.linear_head_helpers import plot_samples
import matplotlib.pyplot as plt
from PIL import Image
add_custom_font("resources/fonts")
import numpy as np

SEED = 100001
torch.manual_seed(SEED)
np.random.seed(SEED)

torch.cuda.empty_cache()
DEVICE = 'cuda:0'
```

```python
FS=60
NS=7
```

```python
%%capture
fig = plot_samples("VOC07", device=DEVICE, num_samples=NS, fs=FS)
fig.savefig("saved/S06_VOC07.jpeg", dpi=300, bbox_inches='tight', pil_kwargs={'optimize': True})
```

```python
%%capture
fig = plot_samples("VOC12", device=DEVICE, num_samples=NS, fs=FS)
fig.savefig("saved/S06_VOC12.jpeg", dpi=300, bbox_inches='tight', pil_kwargs={'optimize': True})
```

```python
%%capture
fig = plot_samples("ADE20K", device=DEVICE, num_samples=NS, fs=FS)
fig.savefig("saved/S06_ADE20K.jpeg", dpi=300, bbox_inches='tight', pil_kwargs={'optimize': True})
```

```python
%%capture
fig_a = Image.open('saved/S06_VOC07.jpeg')
fig_b = Image.open('saved/S06_VOC12.jpeg')
fig_c = Image.open('saved/S06_ADE20K.jpeg')

# Top row
ax1 = plt.subplot(2,2,1)
ax1.imshow(fig_a)
ax1.axis("off")

ax2 = plt.subplot(2,2,2)
ax2.imshow(fig_b)
ax2.axis("off")

# Bottom centered image (span both columns)
ax3 = plt.subplot(2,1,2)
ax3.imshow(fig_c)
ax3.axis("off")

plt.tight_layout()
plt.show()
```
