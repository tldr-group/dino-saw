```python
import torch
from dinosaw.helpers import add_custom_font
from dinosaw.linear_head_helpers import plot_samples
import matplotlib.pyplot as plt
from PIL import Image
add_custom_font("resources/fonts")
import numpy as np

SEED = 100001
torch.manual_seed(SEED)
np.random.seed(SEED)

torch.cuda.empty_cache()
DEVICE = 'cuda:0'
```

```python
FS=45
NS=7
```

```python
# %%capture
fig = plot_samples("VOC07", device=DEVICE, num_samples=NS, fs=FS)
fig.savefig("saved/S06_VOC07.jpeg", dpi=300, bbox_inches='tight', pil_kwargs={'optimize': True})
```

```python
%%capture
fig = plot_samples("VOC12", device=DEVICE, num_samples=NS, fs=FS)
fig.savefig("saved/S06_VOC12.jpeg", dpi=300, bbox_inches='tight', pil_kwargs={'optimize': True})
```

```python
%%capture
fig = plot_samples("ADE20K", device=DEVICE, num_samples=NS, fs=FS)
fig.savefig("saved/S06_ADE20K.jpeg", dpi=300, bbox_inches='tight', pil_kwargs={'optimize': True})
```

```python
%%capture
W, H = 10, 30
FS = 20
add_custom_font('resources/fonts', 'Grotesk')

N_ROWS, N_COLS = 4, 7

fig, axs = plt.subplots(nrows=3, ncols=1, figsize=(W , H))

fig_a = Image.open('saved/S06_VOC07.jpeg')
fig_b = Image.open('saved/S06_VOC12.jpeg')
fig_c = Image.open('saved/S06_ADE20K.jpeg')

axs[0].imshow(fig_a)
axs[0].axis('off')
axs[1].imshow(fig_b)
axs[1].axis('off')
axs[2].imshow(fig_c)
axs[2].axis('off')


labels = ['VOC07', 'VOC12', 'ADE20K']
y_off = [1.01, 1.02, 1.02]
# print(gs.subplots()[0, 0])
for i in range(3):
    ax = axs[i]
    ax.text(-0.01, y_off[i], labels[i], transform=ax.transAxes,
            fontsize=FS+6, fontweight='bold', color='black')

plt.tight_layout()
fig.savefig("saved/S06_combined.jpeg", dpi=300, bbox_inches='tight', pil_kwargs={'optimize': True})
```
