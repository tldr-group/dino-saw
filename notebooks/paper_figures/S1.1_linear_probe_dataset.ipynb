```python
import torch
import matplotlib.pyplot as plt
import numpy as np
from dinosaw.helpers import add_custom_font
import glob
from os import listdir
from PIL import Image

add_custom_font("resources/fonts")

SEED = 100001
torch.manual_seed(SEED)
np.random.seed(SEED)
```

## Microstructures

```python
ds_folder = 'data/linear_probe/homog_micros'
image_files = [f for f in listdir(ds_folder)]
n_imgs = len(image_files)

imgs = []
for img_file in image_files:
    img_path = f'{ds_folder}/{img_file}'
    imgs.append(Image.open(img_path).convert('RGB'))
```

```python
ds_folder = 'data/linear_probe/texture_ds'
image_files = [f for f in listdir(ds_folder)]
np.random.shuffle(image_files)
```

```python
# imgs = []
for img_file in image_files[:15]:
    img_path = f'{ds_folder}/{img_file}'
    imgs.append(Image.open(img_path).convert('RGB'))
```

```python
%%capture
NROW, NCOL = 6, 5
FS = 24
fig, axes = plt.subplots(NROW, NCOL, figsize=(NCOL*1.5, NROW*1.5))
for ax, img in zip(axes.ravel(), imgs):
    ax.imshow(img)
    ax.set_xticks([])
    ax.set_yticks([])


labels = ['(a)', '(b)']
y_off = [1.05, 1.05]
axes = [axes[0, 0], axes[-3, 0]]
# print(gs.subplots()[0, 0])
for i in range(2):
    ax = axes[i]
    ax.text(-0.4, y_off[i], labels[i], transform=ax.transAxes,
            fontsize=FS, fontweight='bold', color='black')


# fig.suptitle("full microstructure dataset used for the linearprobe", fontsize=FS, y=1.02)
fig.tight_layout() #pad=0.05
plt.savefig("saved/S1.1_micro.jpeg", dpi=300, bbox_inches='tight', pil_kwargs={'optimize': True})
```
