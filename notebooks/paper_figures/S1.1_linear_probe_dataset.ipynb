```python
import torch
import matplotlib.pyplot as plt
from os import listdir
from PIL import Image
```

```python
ds_folder = 'data/linear_probe/homog_micros'
image_files = [f for f in listdir(ds_folder)]
n_imgs = len(image_files)

imgs = []
for img_file in image_files:
    img_path = f'{ds_folder}/{img_file}'
    imgs.append(Image.open(img_path).convert('RGB'))
```

```python
%%capture
NROW, NCOL = 3, 5
FS = 50
fig, axes = plt.subplots(NROW, NCOL, figsize=(NCOL*5, NROW*5))
for ax, img in zip(axes.ravel(), imgs):
    ax.imshow(img)
    ax.set_xticks([])
    ax.set_yticks([])
fig.suptitle("Full microstructure dataset used for the linearprobe", fontsize=FS, y=1.02)
fig.tight_layout() #pad=0.05
plt.savefig("saved/S1.1.jpeg", dpi=300, bbox_inches='tight', pil_kwargs={'optimize': True})
```
